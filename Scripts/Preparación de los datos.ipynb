## Organización de los datos:

```python
import geopandas as gpd
import pandas as pd
```

```python
barrios = gpd.read_file("barrio/Barrio.shp")
```

```python
barrios = barrios[~barrios["LIMITECOMU"].isin(["50", "60", "70", "80", "90", "SN01", "SN02"])]
```

```python
df = pd.read_excel("base_area.xlsx")
```

```python
df.head()
```

```python
barrios = barrios.merge(df, left_on="CODIGO", right_on="codigo_barrio", how="left")
```

```python
barrios.head()
```

```python
estrato = gpd.read_file("Estrato/estrato_socioeconomico.shp")
```

```python
estrato = estrato.to_crs(barrios.crs)
```

```python
print(estrato.columns.tolist())
print(estrato.head())
```

```python
# Opción B: moda ponderada por área (más precisa)
estrato["area"] = estrato.geometry.area
estrato_barrio = (
    estrato
    .groupby(["codigo_bar", "estrato"])["area"]
    .sum()
    .reset_index()
    .sort_values("area", ascending=False)
    .drop_duplicates(subset="codigo_bar", keep="first")
    [["codigo_bar", "estrato"]]
    .rename(columns={"estrato": "estrato_moda"})
)
```

```python
# Unir al shape de barrios
barrios = barrios.merge(estrato_barrio, left_on="CODIGO", right_on="codigo_bar", how="left")

# Verificar
print(barrios["estrato_moda"].value_counts())
print(f"Sin estrato: {barrios['estrato_moda'].isna().sum()}")
```

```python
sin_estrato = barrios[barrios["estrato_moda"].isna()][["CODIGO", "geometry"]]
print(sin_estrato[["CODIGO"]])
```

```python
# --- 1. Leer los CSV de GEE ---
ndvi = pd.read_csv("GoogleEarth/ndvi_barrios_2018_2020.csv")
viirs_18 = pd.read_csv("GoogleEarth/viirs_media_barrios_2018.csv")
viirs_19 = pd.read_csv("GoogleEarth/viirs_media_barrios_2019.csv")
viirs_20 = pd.read_csv("GoogleEarth/viirs_media_barrios_2020.csv")
grad_18 = pd.read_csv("GoogleEarth/viirs_gradiente_barrios_2018.csv")
grad_19 = pd.read_csv("GoogleEarth/viirs_gradiente_barrios_2019.csv")
grad_20 = pd.read_csv("GoogleEarth/viirs_gradiente_barrios_2020.csv")

```

```python
# --- 2. Explorar columnas para saber cómo se llama la de barrio y la de valor ---
print(ndvi.columns.tolist())
print(ndvi.head())
```

```python
# --- 1. NDVI ---
barrios = barrios.merge(
    ndvi[["CODIGO", "mean"]].rename(columns={"mean": "ndvi", "CODIGO": "codigo_barrio"}),
    on="codigo_barrio", how="left"
)
```

```python
# --- 2. VIIRS media ---
for year, csv in [(2018, viirs_18), (2019, viirs_19), (2020, viirs_20)]:
    barrios = barrios.merge(
        csv[["CODIGO", "mean"]].rename(columns={"mean": f"viirs_{year}", "CODIGO": "codigo_barrio"}),
        on="codigo_barrio", how="left"
    )
```

```python
# --- 3. VIIRS gradiente ---
for year, csv in [(2018, grad_18), (2019, grad_19), (2020, grad_20)]:
    barrios = barrios.merge(
        csv[["CODIGO", "mean"]].rename(columns={"mean": f"gradiente_{year}", "CODIGO": "codigo_barrio"}),
        on="codigo_barrio", how="left"
    )
```

```python
import numpy as np
from scipy.spatial import cKDTree

# --- 1. Identificar barrios a fusionar ---
# Institucionales (sin estrato) + pequeños
mask_institucional = barrios["estrato_moda"].isna()
mask_pequeno = (barrios["viv_2018"].astype(float) < 500) | (barrios["pob_2018"].astype(float) < 1000)
mask_fusionar = mask_institucional | mask_pequeno

print(f"Institucionales: {mask_institucional.sum()}")
print(f"Pequeños: {(mask_pequeno & ~mask_institucional).sum()}")
print(f"Total a fusionar: {mask_fusionar.sum()}")

# --- 2. Columnas por tipo de operación ---
cols_conteo = ["viv_2018", "viv_2019", "viv_2020",
               "pob_2018", "pob_2019", "pob_2020",
               "h_res_2018", "h_res_2019", "h_res_2020",
               "h_per_2018", "h_per_2019", "h_per_2020",
               "spa_2018", "spa_2019", "spa_2020",
               "cap_spa_2018", "cap_spa_2019", "cap_spa_2020"]

cols_media_ponderada = ["ndvi", "viirs_2018", "viirs_2019", "viirs_2020",
                        "gradiente_2018", "gradiente_2019", "gradiente_2020"]

# --- 3. Asegurar tipos numéricos ---
for c in cols_conteo + cols_media_ponderada + ["SHAPE_AREA"]:
    barrios[c] = pd.to_numeric(barrios[c], errors="coerce")

# --- 4. Encontrar vecino más cercano (por centroide) entre los que NO se fusionan ---
receptores = barrios[~mask_fusionar].copy()
a_fusionar = barrios[mask_fusionar].copy()

centroids_rec = np.array([(g.centroid.x, g.centroid.y) for g in receptores.geometry])
centroids_fus = np.array([(g.centroid.x, g.centroid.y) for g in a_fusionar.geometry])

tree = cKDTree(centroids_rec)
_, indices = tree.query(centroids_fus)

# Asignar el codigo_barrio del receptor
a_fusionar["receptor"] = receptores.iloc[indices]["codigo_barrio"].values

print("\nAsignaciones:")
print(a_fusionar[["codigo_barrio", "NOMBRE", "receptor"]])

# --- 5. Fusionar ---
# Para cada barrio a fusionar, sumar conteos y promediar ambientales ponderado por área
for _, row in a_fusionar.iterrows():
    idx_rec = receptores.index[receptores["codigo_barrio"] == row["receptor"]][0]
    
    # Conteos: suma
    for c in cols_conteo:
        receptores.loc[idx_rec, c] = receptores.loc[idx_rec, c] + (row[c] if pd.notna(row[c]) else 0)
    
    # Ambientales: promedio ponderado por área
    area_rec = receptores.loc[idx_rec, "SHAPE_AREA"]
    area_fus = row["SHAPE_AREA"] if pd.notna(row["SHAPE_AREA"]) else 0
    area_total = area_rec + area_fus
    
    if area_total > 0:
        for c in cols_media_ponderada:
            val_rec = receptores.loc[idx_rec, c] if pd.notna(receptores.loc[idx_rec, c]) else 0
            val_fus = row[c] if pd.notna(row[c]) else 0
            receptores.loc[idx_rec, c] = (val_rec * area_rec + val_fus * area_fus) / area_total
    
    # Área: suma
    receptores.loc[idx_rec, "SHAPE_AREA"] = area_total
    
    # Geometría: unión
    receptores.loc[idx_rec, "geometry"] = receptores.loc[idx_rec, "geometry"].union(row["geometry"])
    
    # Estrato moda: se queda el del receptor (no cambia)

barrios_final = receptores.reset_index(drop=True)
barrios_final = gpd.GeoDataFrame(barrios_final, geometry="geometry", crs=barrios.crs)

print(f"\nBarrios originales: {len(barrios)}")
print(f"Barrios finales: {len(barrios_final)}")
```

```python
# --- Paso 1: Reshape a formato largo ---
years = [2018, 2019, 2020]
panels = []

for y in years:
    temp = barrios_final[["codigo_barrio", "NOMBRE", "geometry",
                          f"h_res_{y}", f"h_per_{y}", f"spa_{y}", f"cap_spa_{y}",
                          f"viv_{y}", f"pob_{y}",
                          f"viirs_{y}", f"gradiente_{y}",
                          "ndvi", "estrato_moda"]].copy()
    temp.columns = ["codigo_barrio", "nombre", "geometry",
                     "hurto_res", "hurto_per", "incaut_spa", "capt_spa",
                     "viviendas", "poblacion",
                     "viirs", "gradiente",
                     "ndvi", "estrato_moda"]
    temp["year"] = y
    panels.append(temp)

panel = gpd.GeoDataFrame(pd.concat(panels, ignore_index=True), geometry="geometry", crs=barrios_final.crs)

print(panel.shape)
print(panel.head())
```
